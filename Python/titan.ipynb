```python
data['Sex']= data["Sex"].replace({
    "male":1,"female":0
})

data['Sex'].value_counts()
```

```python
data["SurvivalStatus"] = np.where(
    data["Survived"] == 1, "Survived" , "Not Survived"
)

data['SurvivalStatus'].value_counts()
```

```python
df2 = data.copy()
df2.head()
```

```python
df1  = data[["PassengerId" , "Name"]]
df2  = data[["PassengerId" , "Survived"]]

result = pd.merge(
    df1,df2,on ="PassengerId", how = "inner"
)

pd.merge(df1,df2,how = "inner")
pd.merge(df1,df2,how = "left")
pd.merge(df1,df2,how = "right")
pd.merge(df1,df2,how = "outer")
```
