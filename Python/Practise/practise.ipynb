```python
marks = [24,67,90,56,40,32,5,43,69,89]

passed = 0 
fail = 0 
for m in marks:
    if m >= 40:
        passed += 1
    else:
        fail += 1
print("PASSED STUDENT : " , passed)
print("FAIL STUDENT : " , fail)

```
