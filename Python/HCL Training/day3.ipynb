```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y=[10,20,30,40,50]
plt.figure(figsize = (6,3))
plt.plot(x,y,color="green",marker='o',linestyle="dashed",linewidth=2,markersize=12)
plt.title('My Data')
plt.xlabel('x axis data')
plt.ylabel('y axis data')
plt.show()

```

```python
x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [23,33,41,53,65]
plt.plot(x,y1,label = "sales 2022")
plt.plot(x,y2,label = "sales 2026")
plt.title("Year Sales")
plt.xlabel("months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
x = ['A','B','C','D','E']
y=[10,20,30,40,50]
plt.bar(x,y)
#plt.plot(x,y,color="green",marker='o',linestyle="dashed",linewidth=2,markersize=12)
plt.title('My Data')
plt.show()
```

```python
import random
data=[random.randint(1,10) for _ in range (50)]
plt.hist(data,bins=8)
plt.title("Histogram Example")
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D', 'E']
sales = [12, 23, 64, 56, 45]

plt.figure(figsize=(6, 4))
plt.pie(sales, labels=categories, autopct='%1.1f%%')
plt.title("Pie Chart Example")
plt.show()
```
